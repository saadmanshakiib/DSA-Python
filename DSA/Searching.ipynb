```python
# Binary Search

array = [12, 23, 36, 45, 51, 64, 76, 78, 83, 93]

item = int(input("Enter Item: "))

low = 0
high = len(array) - 1

while low <= high:

    mid = (low + high) // 2

    if array[mid] == item:
        print("Found")
        break

    elif array[mid] < item:
        low = mid + 1

    else:
        high = mid - 1

else:
    print("Not Found")
```

```python
graph = {
    'S': {'A': 1, 'B': 4},
    'A': {'C': 2, 'D': 5},
    'B': {'D': 1},
    'C': {'G': 5},
    'D': {'G': 3},
    'G': {}
}

heuristic = {
    'S': 7,
    'A': 6,
    'B': 4,
    'C': 4,
    'D': 2,
    'G': 0
}


def a_star(start, goal):

    open_list = [(start, 0)]

    g_cost = {start: 0}

    parent = {}

    while open_list:

        open_list.sort(key=lambda x: x[1])

        current, f_cost = open_list.pop(0)

        if current == goal:

            path = []

            while current in parent:
                path.append(current)
                current = parent[current]

            path.append(start)

            return path[::-1], g_cost[goal]

        for neighbor, cost in graph[current].items():

            new_g = g_cost[current] + cost

            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g

                f = new_g + heuristic[neighbor]

                parent[neighbor] = current

                open_list.append((neighbor, f))

    return None


result = a_star('S', 'G')

print("Path:", result[0])
print("Cost:", result[1])
```
